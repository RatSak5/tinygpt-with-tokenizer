```python
import unicodedata

with open("/kaggle/input/datasets/ratulsak/traindataset/TinyStories-train_clean.txt", "r", encoding="utf-8") as f:
    text = f.read()

# Normalize curly quotes to straight quotes to avoid encoding issues entirely
text = text.replace("’", "'").replace("‘", "'").replace("“", '"').replace("”", '"')
text = unicodedata.normalize("NFKC", text)

with open("chunks/clean_data.txt", "w", encoding="utf-8") as f:
    f.write(text)

print("data cleaned")
```

```python
!python main.py --d-emb 384 --heads 6 --n-decoder-layers 8 --vocab-size 8000 --batch-size 128 --block-size 256 --iterations 50000 --warmup-steps 3500 --eval-interval 100 --device cuda --train-path "chunks/clean_data.txt" --save-path "checkpoints/checkpoint3.pt" --optimizer "adamw" --beta2 0.95
```
